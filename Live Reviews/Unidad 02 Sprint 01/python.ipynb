### La función int()

```python
int(lista_strings[0]) ## Esto cambia lo que hay en la primera posición de la lista de un tipo de variable String a un int.
```

Para mapear toda una lista de objetos se puede usar la siguiente función

```python
list(map(int,lista_strings))
```

Aquí se cambian todos los objetos de la lista a tipo int. map aplica la función que le des (en este caso la función de conversión a int) a todos los elementos de la lista.

Es importante destacar que una string también tiene posiciones y también se puede iterar. La primera posición (la posición 0) es el primer caracter del string.
