#### Cómo se hace el constructor de una clase

```python
class User():
    def init(self, nombre:str, saldo:int, checking_account:bool):
        if not isinstance(saldo, (int,float)):
            raise TypeError("El saldo debe ser un número")
        if saldo < 0:
            raise ValueError("El saldo debe ser positivo")
        self.nombre = nombre
        self.saldo = saldo
        self.checking_account = checking_account
usuario = User(nombre=True, saldo=2, checking_account=True)
usuario
```

##### Como se hacen los métodos de clases

```python
def withdraw(self, cantidad:int):
        if cantidad < 0:
            raise ValueError('La cantidad no puede ser negativa')
        if cantidad > self.saldo:
            raise ValueError('No hay suficiente dinero para retirar')
       
        self.saldo -= cantidad
        return f"Nombre: {self.nombre}, saldo actual: {self.saldo}"

def deposit(self, cantidad:int):
        if cantidad < 0:
            raise ValueError('La cantidad no puede ser negativa')
       
        self.saldo += cantidad #self.saldo = self.saldo + cantidad
        return f"Nombre: {self.nombre}, saldo actual: {self.saldo}"

def transfer(self, otro_usuario, transferir:int):
        '''
        Parámetro otro_usuario:User
        '''
        if self.saldo < transferir:
            raise ValueError("No tienes dineros para transferir")
        if otro_usuario.checking_account != True:
        # if otro_usuario.checking_account == False:
        # if not otro_usuario.checking_account: #checking_account = True
            raise ValueError("Esa cuenta no permite recibir transferencias")
       
        self.saldo -= transferir
        otro_usuario.saldo += transferir
        return f"""Nombre de usuario que transfiere: {self.nombre}, saldo actual: {self.saldo} \n
                    Nombre de usuario de la persona que recibe el dinero: {otro_usuario.nombre} y saldo: {otro_usuario.saldo}"""
```
