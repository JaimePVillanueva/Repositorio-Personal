```python
# TU CÓDIGO AQUÍ
def extract_text(resp) -> str | None:
    try:
        return resp['choices'][0]['message']['content']
    except (KeyError, IndexError, TypeError):
        return None
resp_ok = {"choices": [{"message": {"content": "hola"}}]}
resp_bad = {"choices": []}
print(extract_text(resp_ok))
print(extract_text(resp_bad))
```
